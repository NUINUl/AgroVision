```python
from pathlib import Path
import pandas as pd

ruta_tabular = Path("datasets/crop_recomendation/Crop_recommendation.xls")  #[cite: 3, 4]

# 1. Ver los primeros bytes y primeras líneas del archivo
with open(ruta_tabular, "rb") as f:
  raw_bytes = f.read(300)

print("Primeros bytes (repr):", repr(raw_bytes[:120]))

# 2. Intentar cargar según el formato real
df = None
for metodo, fn in [
    ("read_csv (coma)", lambda: pd.read_csv(ruta_tabular)),
    (
        "read_csv (punto y coma)",
        lambda: pd.read_csv(ruta_tabular, sep=";"),
    ),
    ("read_csv (tabulador)", lambda: pd.read_csv(ruta_tabular, sep="\t")),
    (
        "read_excel (xlrd)",
        lambda: pd.read_excel(ruta_tabular, engine="xlrd"),
    ),
    (
        "read_excel (openpyxl)",
        lambda: pd.read_excel(ruta_tabular, engine="openpyxl"),
    ),
    (
        "read_html",
        lambda: pd.read_html(ruta_tabular)[0]
        if pd.read_html(ruta_tabular)
        else None,
    ),
]:
  try:
    candidato = fn()
    if (
        isinstance(candidato, pd.DataFrame)
        and candidato.shape[0] > 0
        and candidato.shape[1] > 1
    ):
      df = candidato
      print(f"Cargado correctamente usando: {metodo}")
      break
  except Exception as e:
    print(f"Intento {metodo} falló: {type(e).__name__}")

if df is not None:
  print("Dimensiones (filas, columnas):", df.shape)
  print("Columnas:", df.columns.tolist())
  display(df.head())
else:
  print(
      "No se pudo parsear automáticamente; revisa qué muestra 'Primeros bytes'"
      " arriba."
  )
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

df = pd.read_csv("datasets/crop_recomendation/Crop_recommendation.xls")  #

print("Shape:", df.shape)
print("\nValores faltantes:")
print(df.isnull().sum())

print("\nBalance de la variable objetivo ('label'):")  #[cite: 5]
vc = df["label"].value_counts()  #[cite: 5]
display(pd.DataFrame({"conteo": vc, "pct": (vc / len(df) * 100).round(2)}))

print("\nEstadísticas descriptivas numéricas:")
display(df.describe().T)

num_cols = ["N", "P", "K", "temperature", "humidity", "ph", "rainfall"]  #[cite: 5]

# Valores atípicos con regla IQR
outliers = []
for col in num_cols:
  q1, q3 = df[col].quantile(0.25), df[col].quantile(0.75)
  iqr = q3 - q1
  lim_inf, lim_sup = q1 - 1.5 * iqr, q3 + 1.5 * iqr
  n_out = int(((df[col] < lim_inf) | (df[col] > lim_sup)).sum())
  outliers.append({
      "variable": col,
      "n_atipicos": n_out,
      "pct_atipicos": round(n_out / len(df) * 100, 2),
      "lim_inf": round(lim_inf, 2),
      "lim_sup": round(lim_sup, 2),
  })
print("\nValores atípicos (IQR):")
display(pd.DataFrame(outliers))

# Histogramas de distribución
df[num_cols].hist(figsize=(14, 7), bins=25)
plt.suptitle("Distribuciones de variables numéricas - Crop Recommendation", y=1.02)
plt.tight_layout()
plt.show()

# Matriz de correlación
plt.figure(figsize=(8, 6))
sns.heatmap(
    df[num_cols].corr(), annot=True, fmt=".2f", cmap="coolwarm", center=0
)
plt.title("Matriz de correlación (N, P, K, clima y pH)")  #[cite: 5]
plt.tight_layout()
plt.show()
```

```python
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd
from PIL import Image
import seaborn as sns

img_dir = Path(
    "datasets/plant_village_image/Plant_leave_diseases_dataset_without_augmentation"
)  #
exts = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}

filas_clases = []
muestras = {}

for sub in sorted(img_dir.iterdir()):
  if sub.is_dir():
    archivos = [
        f for f in sorted(sub.iterdir()) if f.is_file() and f.suffix.lower() in exts
    ]
    filas_clases.append({"clase": sub.name, "n_imagenes": len(archivos)})
    if archivos:
      muestras[sub.name] = archivos[0]

df_imgs = pd.DataFrame(filas_clases).sort_values("n_imagenes", ascending=False)
total_imgs = df_imgs["n_imagenes"].sum()
df_imgs["pct"] = (df_imgs["n_imagenes"] / total_imgs * 100).round(2)

print(f"Total de clases: {len(df_imgs)} | Total de imágenes: {total_imgs}")
print(
    f"Máximo: {df_imgs['n_imagenes'].max()} ({df_imgs.iloc[0]['clase']}) | "
    f"Mínimo: {df_imgs['n_imagenes'].min()} ({df_imgs.iloc[-1]['clase']})"
)
display(df_imgs)

plt.figure(figsize=(10, max(6, len(df_imgs) * 0.28)))
sns.barplot(data=df_imgs, y="clase", x="n_imagenes")
plt.title("Conteo de imágenes por clase (sin aumento)")
plt.xlabel("Número de imágenes")
plt.tight_layout()
plt.show()

clases_12 = list(muestras.keys())[:12]
fig, axes = plt.subplots(3, 4, figsize=(14, 10))
for ax, clase in zip(axes.flat, clases_12):
  im = Image.open(muestras[clase])
  ax.imshow(im)
  ax.set_title(f"{clase[:28]}...\n{im.size[0]}x{im.size[1]}", fontsize=8)
  ax.axis("off")
plt.tight_layout()
plt.show()
```
